# 05 — Model v2: GPU bi-encoder + cross-encoder

Same pipeline as `04_v1_model` (see `src/pipeline.py`), plus two fine-tuned multilingual transformers
(`paraphrase-multilingual-MiniLM-L12-v2`, Apache-2.0) on Kaggle's 2x T4:

1. **Bi-encoder** (contrastive, in-batch negatives) on training-fold pairs → dense top-5 retrieval per
   record, unioned with the key-based candidates; cosine similarity becomes a feature on every pair.
2. **Cross-encoder** on candidate pairs of records held out from LightGBM (hard negatives) → pair score feature.
3. LightGBM with the new features, one-owner assignment, F0.5-optimal threshold, test output.

Any GPU stage that fails is logged and skipped, so a submission is always written.
