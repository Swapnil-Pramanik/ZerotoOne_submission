# 11 — v7: LLM judge on the hardest pairs

Post-processing of a finished run (attached as a Kaggle input via `kernel_sources`): the hardest pairs only — uncertain
best candidates and close second candidates, with a reserved share for France — are judged by an instruction-tuned LLM
that sees the business, up to three records already matched to it, and the candidate. The LLM score is combined with the
base probability (calibrated on the base run's validation cache when it has one) and the decision is re-made.
See `src/v7.py` and `src/llm_judge.py`.

```python
import sys, json
from pathlib import Path
sys.path.insert(0, str(Path.cwd().resolve().parent / "src"))
from config import ARTIFACT_DIR, OUTPUT_DIR
from v7 import V7Config, run
from llm_judge import JudgeConfig
BASE_RUN = Path("/kaggle/input")        # the attached run's outputs are found by searching for test_scores.parquet
cfg, jcfg = V7Config(), JudgeConfig()
cfg, jcfg
```

```python
report = run(BASE_RUN, OUTPUT_DIR, ARTIFACT_DIR, cfg, jcfg)
print(json.dumps(report, indent=2, default=str))
```
